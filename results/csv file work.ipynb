```python
import pandas as pd

# list of files to combine
csv_files = [
    'results_1.csv',
    'results_2.csv',
    'results_3.csv',
    'results_4.csv',
    'results_5.csv',
    'results_6.csv',
    'results_7.csv',
    'results_8.csv',
    'results_9.csv',
    'results_10.csv',
    'results_11.csv',
    'results_12.csv',
    'results_13.csv'
]

# concat them all together
df_master = pd.concat([pd.read_csv(file) for file in csv_files],
                      ignore_index=True)

print(df_master.head())
```

```python
# data cleaning
df_master_cleaned = df_master.drop(['Values', 'TimestampISO', 'Condition', 'IndexA', 'IndexB'], axis=1)

# create new column for chart type
def assign_type(trial):
    if 1 <= trial <= 20:
        return 'Normal'
    elif 21 <= trial <= 40:
        return 'Diverging'
    else:
        return 'Stacked'
    
df_master_cleaned['ChartType'] = df_master_cleaned['TrialNum'].apply(assign_type)

print(df_master_cleaned.head())
```

```python
# save master csv file
df_master_cleaned.to_csv('master_results.csv')
```
